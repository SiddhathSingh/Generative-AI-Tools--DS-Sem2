```python
import nltk
nltk.download("punkt")
nltk.download("punkt_tab")
```

```python
#1 Basic Sentence Segmentation
from nltk.tokenize import sent_tokenize
text = "English is easy to learn. English is a worldwide spoken language."
sentences = nltk.sent_tokenize(text)
for sent in sentences:
  print(sent)
```

```python
#2 Counting the sentences

text = "English is easy to learn. English is a worldwide spoken language."
sentences = nltk.sent_tokenize(text)
print(len(sentences))
```

```python
#3 Display Sentence Number

text = "English is easy to learn. English is a worldwide spoken language! Do you want to learn English?"
sentences = nltk.sent_tokenize(text)
for i, sent in enumerate(sentences, start=1):
  print(f"Sentence {i}: {sent}")
```

```python
#4 Sentence segmentation with different punctuation
text = "English! is easy to learn. English is a worldwide spoken language, Do you want to learn English?"
sentences = nltk.sent_tokenize(text)
for sent in sentences:
 print(sent)
```

```python
#5 Sentence segmentation from multiple lines
text = """English is easy to learn.
English is a worldwide spoken language.
Do you want to learn English?"""
sentences = nltk.sent_tokenize(text)
for sent in sentences:
 print(sent)
```

```python
#6 Sentence segmentation with abbreviations
text = """English is easy to learn.
English is a worldwide spoken language.
Mr. Sharma has a PHD in English."""
sentences = nltk.sent_tokenize(text)
for sent in sentences:
 print(sent)
```

```python
#8 Display sentence number using spacy
text = """English is easy to learn.
English is a worldwide spoken language!
Mr. Sharma has a PHD in English."""
doc = nlp(text)
for i, sent in enumerate(doc.sents, start =1):
  print(f"Sentence No.{i}: {sent.text}")
```

```python
#10 Access Individual Sentences
text = """English is easy to learn.
English is a worldwide spoken language!
Mr. Sharma has a PHD in English."""
doc = nlp(text)
sent = list(doc.sents)
print(f"First Sentence: {sent[0]}")
print(f"Second Sentence: {sent[1]}")
print(f"Third Sentence: {sent[2]}")
```

```python
#11 Sentence segmentation using spacy sentencizer
# spacy can peform simple rule-based sentence segmentation using sentencizer without loading a large language model
text = """English is easy to learn.
English is a worldwide spoken language!
Mr. Sharma has a PHD in English."""
nlp = spacy.blank("en")
nlp.add_pipe("sentencizer")
doc = nlp(text)
for sent in doc.sents:
  print(sent.text)
```

```python
#12 Sentence segmentation + tokenization using spacy

text = """English is easy to learn.
English is a worldwide spoken language!
Mr. Sharma has a PHD in English."""
nlp = spacy.blank("en")
nlp.add_pipe("sentencizer")
doc = nlp(text)
for sent in doc.sents:
  print("\nSentence:",sent.text)
  print("tokens:")
  for token in sent:
    print(token.text)
```

```python
#13 Sentence segmentation + POS tagging

text = """English is easy to learn.
English is a worldwide spoken language!
Mr. Sharma has a PHD in English."""
nlp = spacy.load("en_core_web_sm")
doc = nlp(text)
for sent in doc.sents:
  print("\nSentence:",sent.text)
  print("tokens:")
  for token in sent:
    print(token.text, "->", token.pos_)
```

```python
#14 Sentence segmentation + NER

text = """English is easy to learn.
English is a worldwide spoken language!
Mr. Sharma has a PHD in English."""
nlp = spacy.load("en_core_web_sm")
doc = nlp(text)
for sent in doc.sents:
  print("\nSentence:",sent.text)
  print("tokens:")
  for entity in sent.ents:
    print(entity.text, "->", entity.label_)
```
