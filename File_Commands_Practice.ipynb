**In this Modeul we will learn how to create, mangae and modify a text file inside python using NLP basics.**

```python
#Creating a text file while giving it some text in order to avoid not found error.
%%writefile demo.txt
Hello World!
This is a Demo text File.
```

```python
#opening the text file created under a variable.
my_file= open('demo.txt')
print(my_file)
```

```python
#Correct way to read a text file
my_file.read()
```

```python
#Reading it again
my_file.read()
```

```python
#Since the file's still open and it's content has been read in one go it requires us to shift the reading point back to the 0th index
my_file.seek(0)
my_file.read()
```

```python
#Printing lines into a list of segregated format
my_file.seek(0)
my_file.readlines()
```

```python
#This method requires us to manually close the file
my_file.close()
```

```python
#Reopening the text file to add more content
my_file = open('demo.txt','w+') #The w is for a writting mode however the + changes it to a reading and writing mode
my_file.write("New Line Added")
my_file.seek(0)
my_file.read() #This method overwrites the content already present in the text file.
```

```python
my_file.close()
```

```python
#Opening the file in append and read mode via the 'a+'
my_file = open('demo.txt','a+')
my_file.write("\nThis is a demo text file.") #append method allows adding the content instead of overwritting it.
my_file.seek(0)
my_file.read()
```

```python
my_file.close()
```

```python
#It's also possible to directly write into the file using the append mode
#Directly writing into the file in append mode.
%%writefile -a demo.txt
Hello World /n
Generative AI Test
```

```python
mf=open('demo.txt')
mf.read()
```

```python
#Using the "with open" method to print file content via line index allows automatically closing the file.
with open('demo.txt','r') as txt:
  first_line = txt.readlines()[1]
print(first_line)
```

```python
#Reading the entire file
with open('demo.txt','r') as file:
  data=file.read()
print(data)
```

```python
#Reading a file line by line
with open('demo.txt','r') as file:
  for line in file:
    print(line.strip()) #.strip removes the extra line added by \n
```

```python
#Using readtime
with open('demo.txt','r') as file:
  line1=file.readline()
  line2=file.readline()
  line3=file.readline() #readline reads one line at a time.
print("First line:",line1.strip())
print("second line:",line2.strip())
print("Third line:",line3.strip())
```

```python
#Counting number of lines
with open('demo.txt','r') as file:
  lines=file.readlines()
print(len(lines))
```

```python
#Counting number of words in a text file
with open('demo.txt','r') as file:
  data=file.read()
words=data.split()
print("Number of words:",len(words))
```

```python
#Counting characters
with open('demo.txt','r') as file:
  data=file.read()

print("Number of characters",len(data))
```

```python
#Searching for a word
sw= input("Enter word to search:")
with open('demo.txt','r') as file:
  data=file.read()
if sw.lower() in data.lower():
  print("Word Found")
else:
  print("Word Not Found")
```

```python
#Counting occurences of a word
s_w= input("Enter word to search:")
with open('demo.txt','r') as file:
  data=file.read()
count = data.lower().count(s_w.lower())
print("Occurences:",count)
```

```python
#Creating a file copy/copying contents of a file to another.
#Searching for a word
with open('demo.txt','r') as file:
  data=file.read()
with open('copy.txt','w') as blank:
  blank.write(data)
with open('copy.txt','r') as blank:
  copyread=blank.read()
print("File contents copied successfully.")
print(copyread)
```

```python
#Reading limited characters
with open('demo.txt','r') as file:
  data=file.read(20)
print(data)
```
