```python
#Reading a PDF file using Python
from PyPDF2 import PdfReader
reader=PdfReader("/content/drive/MyDrive/Colab Notebooks/Files/nlp_practice_urban_bees.pdf")
print("Number of Pagers:", len(reader.pages))
#Reading the first page
page=reader.pages[0]
text=page.extract_text()
print(text)
```

```python
#Reading 1 page from a PDF
for page_number, page in enumerate(reader.pages,start=1):
    print(f"\n---Page {page_number}---")
    text=page.extract_text()
    print(text)
```

```python
#Reading all pages from a PDF with Page Number markers
for page_number, page in enumerate(reader.pages):
    print(f"\n---Page {page_number+1}---")
    text=page.extract_text()
    print(text)
```

```python
#searching for a word in a PDF
search_word="Pollinator"
for page_number, page in enumerate(reader.pages,start=1):
    text=page.extract_text()
    if search_word.lower() in text.lower():
        print(f"Search word '{search_word}' found on page {page_number}")
```

```python
#Counting words in a PDF
word_count=0
for page in reader.pages:
    text=page.extract_text()
    if text:
      word_count += len(text.split())
print("Total Words",word_count)
```

```python
#Copy the first words into another PDf
#Creating a new PDf
from PyPDF2 import PdfWriter
writer=PdfWriter()
writer.add_page(reader.pages[0])
with open ("First_Page_PDF",'wb') as output:
  writer.write(output)
print("Copied.Succesfully")
```
